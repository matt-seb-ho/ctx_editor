```python
import json
from pathlib import Path
from ctx_editor.paths import PROJECT_ROOT
```

```python
def read_json(path):
    with open(path, "r") as f:
        data = json.load(f)
    return data
```

```python
# target runs
# - - path: 2026-01-29/08-03-51
# - - path: 2026-01-29/08-27-43

data_file = PROJECT_ROOT / "data/lic_eval_subset.json"
baseline_run_path = PROJECT_ROOT / "outputs/2026-01-29/08-03-51"
reflect_run_path = PROJECT_ROOT / "outputs/2026-01-29/08-27-43"
```

```python
underlying_data = read_json(data_file)
underlying_dict = {item["task_id"]: item for item in underlying_data}
```

```python
def add_full_spec_and_ground_truth_metadata(run_path: Path):
    for task_dir in (run_path/ "traces").iterdir():
        if not task_dir.is_dir():
            continue
        task_dir_contents = list(task_dir.iterdir())
        assert len(task_dir_contents) == 1
        exp_dir = task_dir_contents[0]
        for trace_file in exp_dir.iterdir():
            if not trace_file.name.endswith(".json"):
                continue
            trace_data = read_json(trace_file)
            task_id = trace_data["sample_id"]
            underlying_item = underlying_dict[task_id]
            metadata = {
                "full_spec_q": underlying_item["full_spec_q"],
                "ground_truth_a": underlying_item["ground_truth_a"],
            }
            trace_data["metadata"] = metadata
            with open(trace_file, "w") as f:
                json.dump(trace_data, f, indent=2)
```

```python
add_full_spec_and_ground_truth_metadata(baseline_run_path)
```

```python
def add_full_spec_and_ground_truth_metadata(run_path: Path):
    results = read_json(run_path / "results.json")
    for item in results:
        task_id = item["sample_id"]
        underlying_item = underlying_dict[task_id]
        item["metadata"]["full_spec_q"] = underlying_item["full_spec_q"]
        item["metadata"]["ground_truth_a"] = underlying_item["ground_truth_a"]
    with open(run_path / "results.json", "w") as f:
        json.dump(results, f, indent=2)
```

```python
add_full_spec_and_ground_truth_metadata(baseline_run_path)
```
